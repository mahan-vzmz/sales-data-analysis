# Sales Data — Initial Inspection

This notebook loads the raw sales dataset and examines its structure, data types, missing values, duplicates, and obvious data-quality issues.

**Important:** No cleaning or modification is performed in this notebook.

## 1. Import Libraries

`Path` manages file paths and `pandas` provides the DataFrame tools used for inspection.

```python
from pathlib import Path

import pandas as pd
```

## 2. Locate the Dataset

The path logic works when the notebook starts from either the project root or the `notebooks` directory.

```python
PROJECT_ROOT = Path.cwd()

if not (PROJECT_ROOT / "data").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_PATH = PROJECT_ROOT / "data" / "raw" / "sales_data.xlsx"

print("Project root:", PROJECT_ROOT)
print("Dataset path:", DATA_PATH)
print("Dataset exists:", DATA_PATH.exists())
```

## 3. Load the Excel File

`pd.read_excel()` reads the worksheet and returns a Pandas DataFrame.

```python
df = pd.read_excel(
    DATA_PATH,
    sheet_name="Raw_Sales_Data",
)

print("Dataset loaded successfully.")
```

## 8. Exact Duplicate Rows

```python
duplicate_count = df.duplicated().sum()
print("Exact duplicate rows:", duplicate_count)
```

## 9. Inspect City and Quantity Values

```python
df["City"].value_counts(dropna=False)
```

```python
df["Quantity"].value_counts(dropna=False)
```

## 10. Inspect Invalid Numeric Values

```python
invalid_prices = df[df["Unit_Price_USD"] <= 0][
    ["Order_Line_ID", "Product", "Unit_Price_USD"]
]

print("Invalid unit prices:", len(invalid_prices))
invalid_prices
```

```python
invalid_discounts = df[
    ~df["Discount_Rate"].between(0, 1)
][
    ["Order_Line_ID", "Product", "Discount_Rate"]
]

print("Invalid discount rates:", len(invalid_discounts))
invalid_discounts
```

## Initial Findings

- The dataset contains 1,295 rows and 11 columns.
- There are 800 unique orders.
- Eight exact duplicate rows exist.
- Some city and payment-method values are missing.
- City names contain whitespace and capitalization inconsistencies.
- Quantity contains invalid numeric and text values.
- Some unit prices are zero or negative.
- Some discount rates are outside the valid range.
- Some order dates are invalid or outside 2025.
- Some products are assigned to incorrect categories.
- No data was cleaned or modified in this notebook.
