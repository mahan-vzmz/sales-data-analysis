## 11. Formal Data-Quality Checks

This section validates the dataset against the rules defined in the data dictionary. No source data is modified.

### 11.1 Validate the Column Structure

```python
EXPECTED_COLUMNS = [
    "Order_Line_ID",
    "Order_ID",
    "Order_Date",
    "Customer_ID",
    "Product",
    "Category",
    "City",
    "Quantity",
    "Unit_Price_USD",
    "Discount_Rate",
    "Payment_Method",
]

missing_columns = [
    column
    for column in EXPECTED_COLUMNS
    if column not in df.columns
]

unexpected_columns = [
    column
    for column in df.columns
    if column not in EXPECTED_COLUMNS
]

print("Missing columns:", missing_columns)
print("Unexpected columns:", unexpected_columns)
```

### 11.2 Define Valid Reference Values

```python
VALID_CITIES = [
    "New York",
    "Chicago",
    "Austin",
    "Seattle",
    "Boston",
    "San Francisco",
]

PRODUCT_CATEGORY_MAP = {
    "Laptop": "Computers",
    "Monitor": "Computers",
    "Keyboard": "Accessories",
    "Wireless Mouse": "Accessories",
    "USB-C Hub": "Accessories",
    "Webcam": "Accessories",
    "Headphones": "Audio",
    "Bluetooth Speaker": "Audio",
    "External SSD": "Storage",
    "USB Flash Drive": "Storage",
    "Office Chair": "Furniture",
    "Desk Lamp": "Furniture",
}
```

### 11.3 Validate Quantity

```python
quantity_numeric = pd.to_numeric(
    df["Quantity"],
    errors="coerce",
)

non_numeric_quantity_count = quantity_numeric.isna().sum()
non_positive_quantity_count = (quantity_numeric <= 0).sum()

print("Non-numeric quantities:", non_numeric_quantity_count)
print("Zero or negative quantities:", non_positive_quantity_count)
```

### 11.4 Validate Order Dates

```python
parsed_dates = pd.to_datetime(
    df["Order_Date"],
    errors="coerce",
)

unparseable_date_count = parsed_dates.isna().sum()
outside_2025_count = (
    parsed_dates.notna()
    & (parsed_dates.dt.year != 2025)
).sum()

print("Unparseable dates:", unparseable_date_count)
print("Dates outside 2025:", outside_2025_count)
```

### 11.5 Validate City Names

```python
non_null_cities = df["City"].dropna()
non_standard_city_count = (
    ~non_null_cities.isin(VALID_CITIES)
).sum()

print("Non-standard city values:", non_standard_city_count)
```

### 11.6 Validate Product and Category Relationships

```python
expected_categories = df["Product"].map(
    PRODUCT_CATEGORY_MAP
)

category_mismatch_count = (
    df["Category"] != expected_categories
).sum()

print("Product/category mismatches:", category_mismatch_count)
```

### 11.7 Build the Quality Summary Table

Each count represents a separate validation rule. Counts should not be added together because one row may fail multiple checks.

```python
quality_summary = pd.DataFrame(
    {
        "Quality_Check": [
            "Exact duplicate rows",
            "Duplicate Order_Line_ID values",
            "Missing city values",
            "Missing payment methods",
            "Non-standard city values",
            "Non-numeric quantities",
            "Zero or negative quantities",
            "Zero or negative unit prices",
            "Discount rates outside 0-1",
            "Unparseable order dates",
            "Dates outside 2025",
            "Product/category mismatches",
        ],
        "Issue_Count": [
            df.duplicated().sum(),
            df["Order_Line_ID"].duplicated().sum(),
            df["City"].isna().sum(),
            df["Payment_Method"].isna().sum(),
            non_standard_city_count,
            non_numeric_quantity_count,
            non_positive_quantity_count,
            (df["Unit_Price_USD"] <= 0).sum(),
            (~df["Discount_Rate"].between(0, 1)).sum(),
            unparseable_date_count,
            outside_2025_count,
            category_mismatch_count,
        ],
    }
)

quality_summary
```
